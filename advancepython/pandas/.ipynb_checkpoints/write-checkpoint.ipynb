```python
import pandas as pd
```

```python
fruits=['Apple','Orange','Banana']
Color=['Red','orange','Yellow']
price=[100,200,300]
basket=zip(fruits,Color,price)
df=pd.DataFrame(basket)
df.columns=["Fruits","Color","Price"]
df
```

```python
df.to_csv()
```

```python
df.to_csv("sample1")
```
