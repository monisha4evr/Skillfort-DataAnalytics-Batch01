```python
import pandas as pd
```

```python
df=pd.read_csv('../../../car_data_500.csv')
df=df.head()
df
```

```python
def convert_to_lower(word:str):
    return word.lower().replace(' ','')
```

```python
convert_to_lower("PANDAS Example")
```

## On Series

```python
df['car'].apply(convert_to_lower)
```

## On DataFrame

```python
def convert_lower(column:pd.Series):
    return column.str.lower()
```

```python
df_ref=df[['car','origin']]
```

```python
df_ref
```

```python
df_ref.apply(convert_lower)
```

```python
def convert_kg_to_tonnes(weight_kg):
    return weight_kg/1000 

def convert_hp_to_kw(hp):
    return hp*0.745
```

```python
df['weight_tones']=df['weight'].apply(convert_kg_to_tonnes)
```

```python
df
```

```python
df['Horsepower'].apply(convert_hp_to_kw)
```

```python
df
```

```python
df.assign(weight_Ton=convert_kg_to_tonnes(df['weight']),
         horsepower_kw=convert_hp_to_kw(df['Horsepower']))
df
```
