# pipe Function
- series/continuos operation
- 1 step then 2nd step ...

```python
def convert_to_kw(hp):
    return hp*0.745

def fix_decimals(value):
    return round(value,2)
```

```python
# transform HOrsepower to kilowatts and fix Decimal
```

```python
df=pd.read_csv('../../../car_data_500.csv').head()
```

```python
df['Kilowatts']=df['Horsepower'].apply(convert_to_kw)
```

```python
df
```

```python
df['kkkwww']=df['Kilowatts'].apply(fix_decimals)
```

```python
df
```

```python
df['pipe_kw_dec']=df['Horsepower'].pipe(convert_to_kw).pipe(fix_decimals)
```
