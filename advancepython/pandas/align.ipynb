```python
import pandas as pd
```

```python
df1=pd.DataFrame([[1,5,9],[5,2,10],[30,90,2]])
df1.columns=['A','B','C']
df1
```

```python
df2=pd.DataFrame([[9,10,5,2],[1,2,3,4],[9,7,4,1],[3,6,6,4]])
df2.columns=['C','D','E','F']
df2
```

```python
# Align Based on Column Value
# align - left,right,inner,outer
```

```python
left, right=df1.align(df2,join='left')
```
