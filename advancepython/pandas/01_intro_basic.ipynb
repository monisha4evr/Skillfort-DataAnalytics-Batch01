# Series

```python
a=[10,20,30,40,50]
myvar=pd.Series(a)
myvar
```

```python
a=[10,20,30,40,50]
myvar=pd.Series(a,index=['a','b','c','d','e'])
myvar
```

```python
a=(10,20,30,40,50)
myvar=pd.Series(a)
myvar
```

```python
a=(10,20,30,40,50)
myvar=pd.Series(a,index=['a','b','c','d','e'])
myvar
```

```python
a={"name":"Apple","color":"Red"}
myvar=pd.Series(a)
myvar
```

```python
a={"name":"Apple","color":"Red"}
myvar=pd.Series(a)
myvar.name="Fruits"
myvar
```

# DataFrames - 2D

# Dataframe using List

```python
suji=["Suji","Chennai","Da"]
vanu=["Vanu","Tanjai","Python"]
swe=["Swetha","Trichy","Java"]
student_details=[suji,vanu,swe]
df=pd.DataFrame(student_details)
df
```

```python
suji=("Suji","Chennai","Da")
vanu=("Vanu","Tanjai","Python")
swe=("Swetha","Trichy","Java")
thiru=("thiru","Salem","AI/Ml")
student_details=[suji,vanu,swe,thiru]
df=pd.DataFrame(student_details)
df
```

# Data frames Using Dictionary

```python
fruits={
    "fruits":['apple','orange','banana'],
    "color":["red","orange","yellow"],
    "price":[100,200,300]
}
df=pd.DataFrame(fruits)
df
```
