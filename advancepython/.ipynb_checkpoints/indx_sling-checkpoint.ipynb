```python
import numpy as np
```

# Indexing and Slicing

```python
a=np.array(np.arange(5))
print(a)
print(a[2])
print(a[-2])
```

```python
a=[[1,2],[3,4]]
print(a)
a=np.array(a)
print(a)
print(a.shape)
print(a.ndim)
print("Oth Row",a[0])
print("1st Row",a[1])
print(a[-1])
print(a[0][1])
print(a[1][0])
print(a[1,0])
```

```python
a=[[[1,2],[3,4]],[[5,6],[7,8]]]
print(a)
a=np.array(a)
print(a)
print(a.shape)
print(a.ndim)
print("0th row",a[0])
print("1st row",a[1])
print("---------------")
print(a[1,0,0])
print("---------------")
print(a[1,1])
```

```python
a=np.array(np.arange(5))
print(a)
print(a[1:4])
print(a[1:])
print(a[:4])
print(a[::-1])
print(a[::2])
```

```python
a=np.array(
    [
        [10,20,30],
        [40,50,60],
        [70,80,90]
    ]
)
print(a)
print("----------")
print(a[0])
print("----------")
print(a[-1])
print("----------")
print(a[::-1])
print("----------")
print(a[:,::-1])
print("----------")
print(a[::-1,::-1])
```
