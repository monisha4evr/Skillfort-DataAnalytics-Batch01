```python
import numpy as np
```

```python
a=[1,2,3,4]
b=[6,7,8,9]
print(a+b)
print(a*b)
```

```python
print(a*3)
print(a+3)
```

```python
a=[1,2,3,4]
b=[6,7,8,9]
a=np.array(a)
b=np.array(b) 
print("Add",a+b)
print("subtract",a-b)
print("Multiply",a*b)
print("divide",a/b)
print("floor division",a//b)
print("modulus",a%b)
print("Exponential",a**2)
```

```python
a=[1,2,3,4]
b=[6,7,8,9]
a=np.array(a)
b=np.array(b) 
print("Add",np.add(a,b))
print("subtract",np.subtract(a,b))
print("Multiply",np.multiply(a,b))
print("divide",np.divide(a,b))
print("floor division",np.floor_divide(a,b))
```

```python
a=[10,20,30,40,55]
a=np.array(a)
print(a*2) # broadcasting
print(a[a%2==0])
print(a%2)
```
