# Attribute - Additional Information 

- ndim [tells Arrays Dimension]
- dtype [ Element data type]
- shape [rows and Columns]
- size [total Element Count]
- itemsize [memory -taken -in bytes]
- nbytes [total memoery  occupied by Elements]
- T (Transpose) Row-> columnn vice versa
- strides (Bytes taken to jump from one element to another)
- flag 

```python
a=np.array([1,2,3,4,5])
print(a.shape)
print(a.size)
print(a.dtype)
print(f"{a.itemsize} bytes")
print(f"{a.nbytes } bytes")
```

```python
a=np.array([1,2,3,4,5],dtype=np.int8)
print(a.shape)
print(a.size)
print(a.dtype)
print(f"{a.itemsize} bytes")
print(f"{a.nbytes } bytes")
```

```python
a=np.array([1,2,3,4,5],dtype=np.float32)
print(a.shape)
print(a.size)
print(a.dtype)
print(f"{a.itemsize} bytes")
print(f"{a.nbytes } bytes")
```

```python
a=np.array([[1,2,3],[4,5,6]])
print(a.shape)
print(a.size)
```

```python
a=np.array([[[1,2,3],[4,5,6]],[[1,2,3],[4,5,6]]])
print(a.shape)
print(a.size)
```

```python
a=np.array([[1,2,3,4,5,6],[6,5,4,3,2,1]])
print("Shape",a.shape)
b=a.T
print(b)
print("Shape",b.shape)
```
