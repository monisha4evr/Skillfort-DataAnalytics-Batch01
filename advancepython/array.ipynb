# Array 
    collection of homogenous values Stored in  a single Variable 
    Stores Data in Same Type, so it is Very Faster,
    used for Mathematical operations.
    
    Type: 
        1D - Single Dimension # only Rows 
        2D - Two Dimension  # Rows and Columns 
        3D - Three Dimension # Base,Rows and Columns
        nD - Multi Dimension # Collection of 2D arrays


```python
import numpy as np
```

```python
a=[1,2,3,4] # 1d 
a= np.array(a)
print(a.ndim)
print(a)
```

```python
a=np.array([[1,2],[3,4]])
a= np.array(a)
print(a.ndim)
print(a)
```

```python
a=np.array([[[1,2],[3,4]],[[1,2],[3,4]]])
a= np.array(a)
print(a.ndim)
print(a)
```
