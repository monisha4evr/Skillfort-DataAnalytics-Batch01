# Numpy
##  NumPy stands for Numerical Python.
##  It is a Python library used for numerical and scientific computing.
##  It provides a powerful N-dimensional array (ndarray) object.
##  It is faster than Python lists because it is implemented in C.
##  It is widely used in Data Analytics, Data Science, Machine Learning, and AI.
##  It integrates well with libraries like Pandas, Matplotlib, and Scikit-learn.

# Features
- Fast compared to Python List 
- Mathematical  and array matrix 
- vectorization
- Higly Used in Data Analytics , Data science , AI/ML
- Less memory usage
- Random number generation
- Array reshaping and manipulation
- Broadcasting support
- Easy integration with other Python libraries

#dataType 
- int8	8-bit integer	(-128 to 127)
- int16	16-bit integer	(eg:1000)
- int32	32-bit integer	(eg:50000)
- int64	64-bit integer	(Very large integers)
- uint8	Unsigned 8-bit integer	(0 to 255)
- float16	16-bit (3–4 decimal digits)	(3.14)
- float32	32-bit (7 decimal digits)	(2.5)
- float64	64-bit (15–16 decimal digits) (default)	(10.25)
- bool_	Boolean	(True, False)
- complex64	Complex number	2+3j
- complex128	High-precision complex	5+6j
- str_	String	"Apple"
- bytes_	Byte string	b"abc"
- object_	Python objects	List, Dict, Custom objects
