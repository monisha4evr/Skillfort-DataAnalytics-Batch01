```python
import matplotlib.pyplot as plt
```

# Subplot
- A subplot allows you to put multiple smaller charts inside a single window (like a grid or a photo collage), 
- instead of opening different windows for each chart
- The three numbers inside plt.subplot(A, B, C)
- mean:A: Total number of rows in your grid layout.
- B: Total number of columns in your grid layout.
- C: The index number of the specific chart you are drawing right now (counts from left to right, top to bottom).

```python
# Sample Data
x = [1, 2, 3, 4,5,6,7]
y = [10, 20, 25, 30,40,50,45]

# --- Chart 1 (Left Side) ---
# plt.subplot(rows, columns, chart_number)
plt.subplot(1, 2, 2)  
plt.plot(x, y, color='red')
plt.title('Line Chart')

# --- Chart 2 (Right Side) ---
plt.subplot(1, 2, 1)  
plt.scatter(x, y, color='blue')
plt.title('Scatter Plot')

# Display everything cleanly
plt.tight_layout()
plt.show()
```

```python
x = [1, 2, 3, 4]
y = [10, 20, 25, 30]

# --- Chart 1 (Top) ---
plt.subplot(2, 1, 1)  
plt.bar(x, y, color='green')
plt.title('Top Chart (Bar)')

# --- Chart 2 (Bottom) ---
plt.subplot(2, 1, 2)  
plt.plot(x, y, color='orange')
plt.title('Bottom Chart (Line)')

plt.tight_layout()
plt.show()
```
