```python
import matplotlib.pyplot as plt
import numpy as np
```

# Boxplot
- Schools & Education: To compare exam scores between different classrooms
- (e.g., Class A vs. Class B vs. Class C). You can instantly see which class performed more consistently.

```python
# 1. 100 Men's height
np.random.seed(42)
heights_men = np.random.normal(175, 7, 100)
plt.boxplot(heights_men)
plt.title('Boxplot of Men\'s Heights')
plt.ylabel('Height (cm)')
plt.show()
```

```python
# 1. 100 Men's height
np.random.seed(42)
heights_men = np.random.normal(175, 7, 100)
plt.boxplot(heights_men, patch_artist=True, boxprops=dict(facecolor='lightblue'))
plt.title('Boxplot of Men\'s Heights')
plt.ylabel('Height (cm)')


plt.show()
```
