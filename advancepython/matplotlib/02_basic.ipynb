```python
import matplotlib.pyplot as plt
import numpy as np
```

# Line Chart

```python
x=[2021,2022,2023,2024,2025,2026]
y=[10000,50000,100000,150000,200000,250000]
plt.plot(x,y)
plt.show()
```

```python
x=np.array([2021,2022,2023,2024,2025,2026])
y=np.array([10000,50000,100000,150000,200000,250000])
plt.plot(x,y)
plt.show()
```

```python
x=np.array([2021,2022,2023,2024,2025,2026])
y=np.array([10000,50000,100000,150000,200000,250000])
plt.plot(x,y,color="pink")
plt.show()
```

```python
x=np.array([2021,2022,2023,2024,2025,2026])
y=np.array([10000,50000,100000,150000,200000,250000])
plt.plot(x,y,color="purple",marker="^") # o-circle s-square ^ Triangle
plt.show()
```

```python
x=np.array([2021,2022,2023,2024,2025,2026])
y=np.array([10000,50000,100000,150000,200000,250000])
plt.plot(x,y,color="purple",
        marker="^",  # o-circle s-square ^ Triangle
        markerfacecolor="yellow",
        markersize=20,
        markeredgecolor="blue",
        markeredgewidth=1,
        linestyle="dashdot", # dashed --  dotted dashdot
        linewidth=2
        )
plt.show()
```

```python
plt.figure(colour="gray")
x=np.array([2021,2022,2023,2024,2025,2026])
y=np.array([10000,50000,100000,150000,200000,250000])
z=np.array([11000,51000,100000,190000,220000,255000])
plt.plot(x,y,color="purple",
        marker="^",  # o-circle s-square ^ Triangle
        markerfacecolor="yellow",
        markersize=20,
        markeredgecolor="blue",
        markeredgewidth=1,
        linestyle="dashdot", # dashed --  dotted dashdot
        linewidth=2,
        label="Branch1" 
        )
plt.plot(x,z,color="Green",
        marker="s",  # o-circle s-square ^ Triangle
        markerfacecolor="skyblue",
        markersize=10,
        markeredgecolor="red",
        markeredgewidth=1,
        linestyle="dotted", # dashed --  dotted dashdot
        linewidth=2,
        label="Branch2" 
        )
plt.legend()
plt.show()
```

```python
plt.plot(y)
plt.show()
```

```python
plotpros=dict(marker="o",  # o-circle s-square ^ Triangle
        markerfacecolor="brown",
        markersize=10,
        markeredgecolor="orange",
        markeredgewidth=1,
        linestyle="dotted", # dashed --  dotted dashdot
        linewidth=2)
```

```python
plt.plot(x,y,color="pink",**plotpros)
plt.show()
```

```python
plt.xlabel("Year" ,color="Blue")
plt.ylabel("Revenue",color="Green")
plt.plot(x,y , color="skyblue")
plt.grid()
plt.show()
```
