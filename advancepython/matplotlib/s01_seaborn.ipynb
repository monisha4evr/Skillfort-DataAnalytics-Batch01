# Seaborn 
- Python data visualization library built on top of Matplotlib. 
- It helps us create beautiful, attractive, and informative statistical charts with less code.

# installation
- **pip install seaborn** -> for env -> python -m pip install seaborn
- (Reason for python -m =>its takes the current active python -from Virtual Environment rather then Original python)

# Why do we use Seaborn?
- Creates attractive graphs with default colors and themes.
- Works directly with Pandas DataFrames.
- Best for data analysis and statistical visualization.
- Requires less code than Matplotlib.

# Seaborn vs Matplotlib
- Matplotlib - Seaborn
- Basic plotting library. - Built on top of Matplotlib.
- More code required. - Less code required.
- Simple chart design. - Beautiful default themes and colors.
- General-purpose visualization.- Statistical and analytical visualization.

```python
sns.displot(tips)
```

```python
sns.jointplot(x='total_bill',y='tip',data=tips,hue='sex')
```

```python
sns.jointplot(x='total_bill',y='tip',data=tips,kind='reg')
```

```python
sns.jointplot(x='total_bill',y='tip',data=tips,kind='kde')
```

```python
sns.set_style('ticks') # ticks,white
sns.jointplot(x='total_bill',y='tip',data=tips, height=5)
```

```python
sns.pairplot(tips)
```

```python
sns.barplot(x='sex',y='total_bill',data=tips,)
```

```python
sns.boxplot(x='day',y='total_bill',data=tips,hue='sex')
plt.legend(loc=1)
```

```python
plt.figure(figsize=(5,5))
sns.violinplot(x='day',y='total_bill',data=tips,hue='sex') #- ,split=True
plt.legend(loc=0)
```

```python
sns.violinplot(x='day',y='total_bill',data=tips,hue='sex',split=True) 
```

```python
sns.swarmplot(x='day',y='total_bill',data=tips,hue='sex',palette='dark:blue')
```

```python
sns.stripplot(x='day',y='total_bill',data=tips,hue='sex',palette='winter',jitter=True,dodge=True)
```

```python
sns.pairplot(tips,hue='sex')
```
