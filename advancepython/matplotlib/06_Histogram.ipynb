```python
import matplotlib.pyplot as plt
import numpy as np
```

# Histogram
- A histogram looks like a bar chart, but it is used to show how often data numbers fall into specific ranges (called "bins"). 
- For example, it can show how many people in a room are between 10–20 years old, 20–30 years old, and 30–40 years old.

```python
# 1. Collect your data numbers
exam_scores = [65, 72, 78, 82, 85, 88, 91, 92, 95, 99, 45, 55, 74, 83, 86]

# 2. Create the histogram 
# 'bins' splits the data into ranges. 'edgecolor' adds lines between bars.
plt.hist(exam_scores, bins=[40, 50, 60, 70, 80, 90, 100], color='skyblue', edgecolor='blue')

# 3. Add labels
plt.title('Distribution of Exam Scores')
plt.xlabel('Score Ranges')
plt.ylabel('Number of Students (Count)')

# 4. Show the chart
plt.show()
```

```python
# Generate 1000 random data points (Normal Distribution)
np.random.seed(42)
heights_men = np.random.normal(175, 7, 500)
heights_women = np.random.normal(162, 6, 500)
#center_value=Mean
#Spread_Range= STD
# np.random.normal( Center_Value, Spread_Range, Total_Numbers_Count )

# Plot two overlapping histograms with transparency (alpha)
plt.hist(heights_men, bins=20, alpha=0.6, color='blue', edgecolor='darkblue', label='Men')
plt.hist(heights_women, bins=20, alpha=0.6, color='pink', edgecolor='darkred', label='Women')

# Labels
plt.title('Comparison of Heights Distribution')
plt.xlabel('Height (cm)')
plt.ylabel('Frequency')

# Move the legend OUTSIDE on the right side
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left', borderaxespad=0)

# Tight layout stops the right side from cutting off
plt.tight_layout()
plt.show()
```
