# Matplotlib
## Matplotlib is an open-source, low-level data visualization library in Python used for creating static, interactive, and animated plots
- Installation 
- pip install matplotlib
- To Start : import matplotlib
## pyplot
- What is Pyplot?matplotlib.pyplot is a collection of functions that make Matplotlib work like MATLAB. 
- Each pyplot function makes some change to a figure
- (e.g., creates a figure, creates a plotting area, plots lines, or decorates the plot with labels
- To Create: import matplotlib.pyplot as plt

```python
import matplotlib.pyplot as plt
```

```python
1. Figure plt.figure(figsize=(width, height), dpi=dots_per_inch, facecolor=background_color)
2. plot
- plt.figure(): Creates a new workspace canvas.
- plt.plot(): Draws line plots or markers.
- plt.show(): Displays the final visual output.
- plt.title() / plt.xlabel(): Adds text descriptors.
- plt.grid(): Toggles background grid lines.
- plt.clf(): Clears the current active figure
- plt.legend(): adds a visual guide to charts based on Lable: eg: label='Ascending'
```

```python
import matplotlib.pyplot as plt

# 1. Plot data and assign labels
plt.plot([1, 2, 3], [2, 4, 6], label='Ascending')
plt.plot([1, 2, 3], [6, 4, 2], label='Descending')

# 2. Render the legend

plt.show()
```
