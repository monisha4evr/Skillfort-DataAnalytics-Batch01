```python
import matplotlib.pyplot as plt
import numpy as np
```

# Scatter Chart
- show the relationship between two different things
- eg: Study Hours vs Marks

```python
# 1. Define data points (e.g., Study Hours vs. Exam Grades)
hours_studied = [1, 2, 3, 4, 5, 5.5, 6, 7, 8, 9, 10]
exam_grades = [55, 62, 65, 71, 74, 78, 85, 82, 89, 95, 99]

# 2. Create the scatter plot
plt.scatter(hours_studied, exam_grades, color='blue', marker='o')

# 3. Add titles and labels
plt.title('Relationship Between Study Hours and Grades')
plt.xlabel('Hours Studied')
plt.ylabel('Exam Grade (%)')

# 4. Display the plot
plt.grid(True)
plt.show()
```

```python
# Seed for reproducibility
np.random.seed(42)

# Generate 50 random data points
n = 50
x = np.random.rand(n) * 100        # X coordinates
y = np.random.rand(n) * 100        # Y coordinates
colors = np.random.rand(n) * 100   # Maps to a color spectrum (e.g., Temperature)
sizes = np.random.rand(n) * 500     # Maps to marker sizes (e.g., Population)

# Create the advanced scatter plot
plt.scatter(x, y, c=colors, s=sizes, alpha=0.6, cmap='viridis', edgecolors='black')

# Add a colorbar legend to interpret the color values
plt.colorbar(label='Intensity Scale')

# Labels and layout
plt.title('Advanced Scatter Plot (4D Data Visualization)')
plt.xlabel('X Coordinate')
plt.ylabel('Y Coordinate')

plt.show()
```
