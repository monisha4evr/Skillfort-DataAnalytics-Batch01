# Seaborn 
- Python data visualization library built on top of Matplotlib. 
- It helps us create beautiful, attractive, and informative statistical charts with less code.

# installation
- **pip install seaborn** -> for env -> python -m pip install seaborn
- (Reason for python -m =>its takes the current active python -from Virtual Environment rather then Original python)

# Why do we use Seaborn?
- Creates attractive graphs with default colors and themes.
- Works directly with Pandas DataFrames.
- Best for data analysis and statistical visualization.
- Requires less code than Matplotlib.

# Seaborn vs Matplotlib
- Matplotlib - Seaborn
- Basic plotting library. - Built on top of Matplotlib.
- More code required. - Less code required.
- Simple chart design. - Beautiful default themes and colors.
- General-purpose visualization.- Statistical and analytical visualization.
