```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
x=np.array(["video","shorts","live","post"])
y=np.array([3060,2900,4000,2000])
plt.pie(y,labels=x)
plt.title("Sample Pie Diagram",color="Red")
plt.show()
```

```python
x=np.array(["video","shorts","live","post"])
y=np.array([3060,2900,4000,2000])
plt.pie(y,labels=x,
         colors=['purple','blue','orange','pink'],
       )
plt.title("Sample Pie Diagram",color="Red")
plt.show()
```

```python
x=np.array(["video","shorts","live","post"])
y=np.array([3060,2900,4000,2000])
plt.pie(y,labels=x,
       autopct="%1f%%",#to show % give 2 times %%
       colors=['purple','blue','orange','pink'],
       )
plt.title("Sample Pie Diagram",color="Red")
plt.show()
```

```python
x=np.array(["video","shorts","live","post"])
y=np.array([3060,2900,4000,2000])
plt.pie(y,labels=x,
       autopct="%1f%%",#to show % give 2 times %%
       colors=['purple','blue','orange','pink'],
       shadow=True 
       )
plt.title("Sample Pie Diagram",color="Red")
plt.show()
```

```python
x=np.array(["video","shorts","live","post"])
y=np.array([3060,2900,4000,2000])
plt.pie(y,labels=x,
       autopct="%1f%%",#to show % give 2 times %%
       colors=['purple','blue','orange','pink'],
       shadow=True,
       explode=[0,0.5,0,0], #to take that piece out
       startangle=90   
       )

plt.title("Sample Pie Diagram",color="Red")
plt.show()
```

```python
x=np.array(["video","shorts","live","post"])
y=np.array([3060,2900,4000,2000])
plt.pie(y,labels=x,
       autopct="%1f%%",#to show % give 2 times %%
       colors=['purple','blue','orange','pink'],
       shadow=True,
       explode=[0,0.5,0,0], #to take that piece out
       startangle=180   
       )

plt.title("Sample Pie Diagram",color="Red")
plt.legend()
plt.show()
```

```python
x=np.array(["video","shorts","live","post"])
y=np.array([3060,2900,4000,2000])
plt.pie(y,labels=x,
       autopct="%1f%%",#to show % give 2 times %%
       colors=['purple','blue','orange','pink'],
       shadow=True,
       explode=[0,0.5,0,0], #to take that piece out
       startangle=180   
       )

plt.title("Sample Pie Diagram",color="Red")
# ---- THIS IS THE FIX TO MOVE THE LEGEND OUTSIDE RIGHT ----
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left', borderaxespad=0)
# Ensures the legend does not get cut off when saving or displaying
plt.tight_layout() 
plt.show()
```

```python
x=np.array(["video","shorts","live","post"])
y=np.array([3060,2900,4000,2000])
plt.pie(y,labels=x,
       autopct="%1f%%",#to show % give 2 times %%
       colors=['purple','blue','orange','pink'],
       shadow=True,
       explode=[0,0.5,0,0], #to take that piece out
       startangle=180   
       )

plt.title("Sample Pie Diagram",color="Red")
plt.legend(bbox_to_anchor=(1.15, 0.5), loc='center left')
plt.show()
```
