```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
x=np.array([2021,2022,2023,2024,2025,2026])
y=np.array([10000,50000,100000,10000,200000,50000])
plt.bar(x,y,color="green")
plt.xlabel("year",color="pink")
plt.ylabel("Revenue")
plt.show()
```

```python
x=np.array([2021,2022,2023,2024,2025,2026])
y=np.array([10000,50000,100000,10000,200000,50000])
plt.barh(x,y,color="green")
plt.xlabel("year",color="pink")
plt.ylabel("Revenue")

plt.show()
```
